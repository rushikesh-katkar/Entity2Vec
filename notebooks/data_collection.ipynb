```python
import wikipedia
import re

all_docs = []
visited = set()

def clean_key(key):
    return re.sub(r'[^a-z]', '', key.lower())

def get_wiki_categories(page_title):
    """Fetch categories of a Wikipedia page (used as topic)."""
    try:
        page = wikipedia.page(page_title)
        return page.categories  # list of categories
    except Exception as e1:

        try:
            results = wikipedia.search(page_title)
            page = wikipedia.page(results[0], auto_suggest=False)
            return page.categories
        
        except Exception as e2:
            print(f"Error fetching page {page_title}: {e2}")
            return []


def scan_page(start_list, movie_tokens, scientist_tokens, bridge_tokens):
    """Scan a Wikipedia page using only its topic (categories)."""

    ct = 0
    global visited, all_docs

    queue_movie      = [start_list[1]]
    queue_scientists = [start_list[0]]
    queue_bridge     = []

    prev = "scientist"

    while ((len(queue_movie) > 0) | (len(queue_scientists) > 0) | (len(queue_bridge) > 0)):

        if prev == 'scientist':
            prev = "bridge"
            if len(queue_bridge) > 0:
                current = queue_bridge.pop(0)
            else:
                continue

        elif prev == 'bridge':
            prev = "movie"
            if len(queue_movie) > 0:
                current = queue_movie.pop(0)
            else:
                continue

        elif prev == 'movie':
            prev = "scientist"
            if len(queue_scientists) > 0:
                current = queue_scientists.pop(0)
            else:
                continue

        if current in visited:
            continue
        visited.add(current)

        categories = [c.lower() for c in get_wiki_categories(current)]
        doc = []
        doc_links = []

        try:
            page = wikipedia.page(current)
            links = page.links  # links only on this page
        except Exception as e:
            print(f"Error fetching links for {current}: {e}")
            links = []

        for link in links:
            cleaned_link = clean_key(link)
            if cleaned_link in movie_tokens:

                movie_keywords = ["film", "movie", "cinema", "feature film", "films", "directed by", "filmography"]
                if any(keyword in cat for cat in categories for keyword in movie_keywords):
                    doc.append(movie_tokens[cleaned_link])
                    doc_links.append(link)
                    queue_movie.append(link)

            elif cleaned_link in scientist_tokens:
                
                if any(keyword in cat for cat in categories for keyword in ["scientist", "physicist", "researcher", "chemist", "biologist"]):
                    doc.append(scientist_tokens[cleaned_link])
                    doc_links.append(link)
                    queue_scientists.append(link)

            elif cleaned_link in bridge_tokens:
                
                doc.append(bridge_tokens[cleaned_link])
                doc_links.append(link)
                queue_bridge.append(link)

        ct += 1

        if ct == 500:
            break
                
        all_docs.append({'doc': doc, 'doc_links': doc_links})
```

```python
scan_page(["Issac Newton",  "Arrival (2016)"], movie_tokens, scientist_tokens, bridge_tokens)
```

```python
import networkx as nx
import matplotlib.pyplot as plt


G = nx.Graph()

for doc_entry in all_docs:
    tokens_in_doc = doc_entry['doc']
    
    for token in tokens_in_doc:
        if token not in G:
            if token in movie_tokens.values():
                G.add_node(token, type='movie')
            elif token in scientist_tokens.values():
                G.add_node(token, type='scientist')
            else:
                G.add_node(token, type='bridge')
    
    for i in range(len(tokens_in_doc)):
        for j in range(i + 1, len(tokens_in_doc)):
            G.add_edge(tokens_in_doc[i], tokens_in_doc[j])

color_map = {'movie': 'red', 'scientist': 'blue', 'bridge': 'green'}
node_colors = [color_map[G.nodes[n]['type']] for n in G.nodes]
```

```python

plt.figure(figsize=(8, 6))
nx.draw(G, node_color=node_colors, node_size=200, with_labels=False)
plt.title("Token Graph from all_docs (colored by type)")
plt.show()
```
